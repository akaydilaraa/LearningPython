```python
class Person():
    name=""
    age=0

#method, initializer
    #yazılması zorunlu adımlar tanımladık
    def __init__(self, age,name):
        self.name=name
        self.age=age
```

```python
class Person():
    #property kısmı
    name=""
    age=0

#method, initializer
    #yazılması zorunlu adımlar tanımladık
    def __init__(self, age,name):
        self.name=name
        self.age=age

    def test(self):
        print("test")
```

```python
samet=Person(12,"Samet")
```

```python
samet.test()
```
