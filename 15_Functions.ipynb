```python
#return yapabiliyoruz yani çıktı verebiliyoruz
```

```python
def summation(num1, num2, num3):
    print(num1+num2+num3)
```

```python
summation(1,2,3)
```

```python
x=summation(1,2,3)
```

```python
print(x) #çünkü o fonksiyon sadece dönderme işlemi yapabiliyor 
#farklı bir değişkene atanamaz return edilmediği için
```

```python
#bu fonksiyonu tanımladık ama geriye bir şey döndermiyor
#sadece print işlemi var ve o fonksiyonla tanımlı
#return olduğunda ise dışarıya bu çıktıyı verebilecek hale getiriyoruz
```

```python
def return_summation(num1,num2,num3):
    return num1+num2+num3
```

```python
return_summation(1,2,3)
```

```python
x=return_summation(1,2,3)
```

```python
print(x)
```

```python
#args, kwargs (arguments, key word arguments)
```

```python
def args_sum(*args):
    return sum(args)
```

```python
args_sum(10,20,30)
```

```python
#*args yaparsak içine kaç tane değişken alacağını söylemiyoruz
#kaç değişken gelirse al diyoruz
#sum fonksiyonunu kullanarak gelen tüm sayıları topladık
```

```python
#kwargs
```

```python
def kwargs_example(**kwargs):
    print(kwargs)
```

```python
kwargs_example(apple=100, banana=150)
```

```python
#bir dictionary olarak kullanıyoruz 
```

```python
def kwargs_example2(**kwargs):
    if "apple" in kwargs:
        print("Aplleee")
    else :
        print(":(")
        

```

```python
kwargs_example2(banana=10, apple=123)
```

```python
kwargs_example2(banana=10)
```
