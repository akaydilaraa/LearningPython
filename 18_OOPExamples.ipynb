```python
class Musician():

    def __init__(self,name):
        self.name=name
        print("musician class")

    def test1(self):
        print("test")

    def test2(self):
        print("test2")

```

```python
class MusicianPlus(Musician): #musician sınıfınfan kalıtım almak için
    def __init__(self,name):
        Musician.__init__(self,name)
        print("musician plus")

    def test3(self):
        print("Test3")

    def test1(self): #override
        print("TEST1 TEST1")
    
```

```python
#Burada test1 ilk sınıfta tanımlıydı ama ikinci sınıfta kalıtım aldığımız sınıfta
#bu durumu değiştirebiliriz
```

```python
ali=MusicianPlus("Ali")
```

```python
ali.test1()
```

```python
#Bu kısımda kalıtım aldığımız sınıfta bulunan bir metodu değiştirebildik
#Buna override denir
```

```python
dilara.test1()
```



POLYMORPHISM



```python
#Aynı ismi kullanarak farklı işler yapabilen 
```

```python
class Banana():
    def __init__(self,name):
        self.name=name
    def info(self):
        return f"100 calories {self.name}"
```

```python
class Apple():
    def __init__(self,name):
        self.name=name
    def info(self):
        return f"150 calories {self.name}"
```

```python
banana=Banana("banana")
```

```python
apple=Apple("apple")
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruitList= [banana,apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

```python
#farklı sınıflara gidip o sınıflardaki aynı isimli info metodunu bulup
#çalıştırabildi
```

ENCAPSULATION

```python
class Phone1():
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def info(self):
        print(f"{self.name} price is:{self.price}")
```

```python
iphone=Phone1("Iphone 14",500)
```

```python
iphone.info()
```

```python
iphone.price=400
```

```python
iphone.info()
```

```python
class Phone():
    def __init__(self,name,price):
        self.name=name
        #İki tane alt çizgi koyarak price değişkenini sadece bu sınıf içerisinden 
        #erişilebilir hale getirdik
        self.__price=price
    def info(self):
        print(f"{self.name} price is:{self.__price}") 
```

```python
iphonex=Phone("Iphone 15",550)
```

```python
iphonex.__price=200
```

```python
iphonex.info()
```

ABSTRACTION

```python
#SOYUT SINIF OLUŞTURMA 
#özel bir keyword yok ama bir modül tanımlanmış
```

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):

    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
#Tek başına soyut sınıftan bir nesne oluşturamayız ama o abstract sınıftan 
#yeni sınıflar türetir başta tanımladığımız sınıf özelliklerini kullanmayı zorunlu kılarız
```

```python
class Tesla(Car):
    
    def maxSpeed(self):
        print("200km")
```

```python
tesla=Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):
    def maxSpeed(self):
        print("300km")
```

```python
mercedes=Mercedes()
```

```python
mercedes.maxSpeed()
```

```python
class Fruit():
    def __init__(self,name,calories):
        self.name=name
        self.calories=calories

    def __str__ (self):
        return f"{self.name}: {self.calories}"
```

```python
myFruit=Fruit("banana",150)
```

```python
myFruit.calories
```

```python
myFruit.name
```
