```python
my_new_list=[("a","b"),("c","d"), ("e","f")]
```

```python
for element in my_new_list:
    print(element)
```

```python
for (x,y) in my_new_list:
    print(x)
```

```python
my_list=[10,20,30,40,50]
```

```python
for num in my_list:
    print(num)
    if num==40:
        print("yes")
```

```python
for num in my_list:
    print(num)
    if num==40:
        print("yes")
        break
```

```python
my_list=[10,20,30,40,50,60,70]
```

```python
for num in my_list:
    print(num)
    if num==40:
        print("yes")
        continue
```

```python
for num in my_list:
    print(num)
    if num==40:
        
        continue
        print("yes")
```
