```python
#bunu herhangi bir fonksiyon için yapabiliriz
```

```python
def controlString (string):
    return "a" in string
```

```python
controlString("Dilara")
```

```python
controlString("Cem")
```

```python
list(map(controlString,"Ali"))
```

```python
#map() fonksiyonu, verilen iterable (yani üzerinde gezilebilir, döngü kurulabilir) nesnenin her bir elemanını sırayla fonksiyona argüman olarak gönderir.
```

```python
list(map(controlString,"ali"))
```

FILTER

```python
myStringList=["Dilara", "Ahmet", "Cem"]
```

```python
list(filter(controlString, myStringList))
```

KONTROL VERME MEKANİZMASI 
LEGB LOCAL ENCLOSING GLOBAL BUILT-IN

```python
#eğer fonksiyonun içerisinden dışarıda tanımlı olan bir değişkeni değiştirmek istersek
```

```python
y =10
def deneme():
    y=3
    print(y)
```

```python
deneme()
```

```python
y
```

```python
y =10
def deneme2():
    global y
    y=3
    print(y)
```

```python
deneme2()
```
